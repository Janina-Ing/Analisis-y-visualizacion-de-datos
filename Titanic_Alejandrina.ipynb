# PARTE B Aplicar los ejercicios a Titanic

```python
def cargar_titanic(path: str) -> pd.DataFrame:
    """Carga Titanic y hace ajustes básicos."""
    titanic = pd.read_csv(path)

    # Si viene con un índice guardado, lo quitamos
    if "Unnamed: 0" in titanic.columns:
        titanic = titanic.drop(columns=["Unnamed: 0"])

    return titanic


def parte_b_titanic(titanic: pd.DataFrame) -> pd.DataFrame:
    sep("PARTE B: Titanic (repetición de ejercicios con columnas reales)")

    # Validación básica de columnas esperadas
    required_cols = {"survived", "sex", "age", "fare", "embarked"}
    missing = required_cols - set(titanic.columns)
    if missing:
        print("\n[ADVERTENCIA] Faltan columnas esperadas:", missing)
        print("Revisa tu Titanic.csv y ajusta el script a tus columnas reales.\n")

    # B2) Exploración rápida
    sep("B2) Exploración rápida")
    print("shape:", titanic.shape)
    print("\nColumnas:\n", list(titanic.columns))

    print("\ninfo():")
    titanic.info()

    print("\ndescribe(include='all'):")
    # include='all' puede ser muy largo, pero es útil para entrega
    print(titanic.describe(include="all"))

    print("\nNulos por columna (top 10):")
    print(titanic.isnull().sum().sort_values(ascending=False).head(10))

    print("\nValores únicos (top 10):")
    print(titanic.nunique().sort_values(ascending=False).head(10))

    return titanic
```

# Agrupación y agregación

```python
sep("B5) Agrupación y agregación")
if {"sex", "survived"}.issubset(titanic_clean.columns):
    survival_by_sex = titanic_clean.groupby("sex", observed=False)["survived"].mean().sort_values(ascending=False)
    print("\nTasa de supervivencia por sex:\n", survival_by_sex)

if {"pclass_num", "survived"}.issubset(titanic_clean.columns):
    agg_by_class = titanic_clean.groupby("pclass_num", observed=False).agg(
        pasajeros=("survived", "count"),
        tasa_supervivencia=("survived", "mean"),
        edad_prom=("age", "mean") if "age" in titanic_clean.columns else ("survived", "count"),
        fare_mediana=("fare", "median") if "fare" in titanic_clean.columns else ("survived", "mean")
    )
    print("\nAgregaciones por clase:\n", agg_by_class)

if {"sex", "class", "fare"}.issubset(titanic_clean.columns):
    pivot_fare = pd.pivot_table(
        titanic_clean,
        values="fare",
        index="sex",
        columns="class",
        aggfunc="mean",
        observed=False
    )
    print("\nPivot: fare promedio por sex y class:\n", pivot_fare)

if {"sex", "survived"}.issubset(titanic_clean.columns):
    ct = pd.crosstab(titanic_clean["sex"], titanic_clean["survived"])
    print("\nCrosstab sex vs survived:\n", ct)
```
